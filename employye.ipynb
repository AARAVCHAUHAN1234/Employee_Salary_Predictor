```python
def remove_outliers(data,column):
    Q1=data[column].quantile(0.25)

    Q3=data[column].quantile(0.75)

    IQR=Q3-Q1

    lower=Q1-1.5*IQR

    upper=Q3+1.5*IQR

    return data[(data[column]>=lower) & (data[column]<=upper)]
```

```python
for col in numerical_cols:
    if col!="Annual_Salary_LPA":
        data=remove_outliers(data,col)
```

```python
from sklearn.preprocessing import LabelEncoder

le=LabelEncoder()

data['Gender']=le.fit_transform(data['Gender'])

data['Remote_Work']=le.fit_transform(data['Remote_Work'])
```

```python
X=data.drop('Annual_Salary_LPA',axis=1)
y = data["Annual_Salary_LPA"]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
from xgboost import XGBRegressor
```

```python
lr=LinearRegression()
lr.fit(X_train,y_train)
```

```python
rf=RandomForestRegressor()
rf.fit(X_train,y_train)
```

```python
xgb=GradientBoostingRegressor()
xgb.fit(X_train,y_train)
```

```python
xg=XGBRegressor()
xg.fit(X_train,y_train)
```

```python
y_pred1=lr.predict(X_test)
y_pred2=rf.predict(X_test)
y_pred3=xgb.predict(X_test)
y_pred4=xg.predict(X_test)
```

```python
from sklearn import metrics
```

```python
score1=metrics.r2_score(y_test,y_pred1)
score2=metrics.r2_score(y_test,y_pred2)
score3=metrics.r2_score(y_test,y_pred3)
score4=metrics.r2_score(y_test,y_pred4)
```

```python
print(score1,score2,score3,score4)
```

```python
Final_data=pd.DataFrame({'Models':['LR','RF','GBR','XG'],"R2_SCORE":[score1,score2,score3,score4]})
```

```python
Final_data
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
models_predictions = {
    "Linear Regression": y_pred1,
    "Random Forest": y_pred2,
    "Gradient Boosting": y_pred3,
    "XGBoost": y_pred4
}

results = []

for name, pred in models_predictions.items():
    mae = mean_absolute_error(y_test, pred)
    rmse = np.sqrt(mean_squared_error(y_test, pred))
    r2 = r2_score(y_test, pred)

    results.append({
        "Model": name,
        "MAE": mae,
        "RMSE": rmse,
        "R2_SCORE": r2
    })

results_df = pd.DataFrame(results)

results_df.sort_values(by="R2_SCORE", ascending=False)
```
